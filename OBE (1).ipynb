```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


from sklearn.model_selection import (
    train_test_split,
    GridSearchCV,
    StratifiedKFold
)

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
```

```python
# ============================================================
# CHECK MISSING VALUES
# ============================================================

missing_values = obe.isnull().sum()

print(
    missing_values
)

print(
    "\nTotal Missing Values:",
    missing_values.sum()
)
```

```python
# ============================================================
# REMOVE EXACT DUPLICATE ROWS
# ============================================================

print(
    "Duplicate Rows Before:",
    obe.duplicated().sum()
)


obe_cleaned = (
    obe
    .drop_duplicates()
    .reset_index(drop=True)
    .copy()
)


print(
    "Duplicate Rows After:",
    obe_cleaned.duplicated().sum()
)


print(
    "Dataset Shape:",
    obe_cleaned.shape
)
```

```python
# ============================================================
# CHECK INVALID PHYSICAL VALUES
# ============================================================

invalid_mask = (

    (obe_cleaned['Age'] <= 0)

    |

    (obe_cleaned['Height'] <= 0)

    |

    (obe_cleaned['Weight'] <= 0)

)


print(
    "Invalid Rows:",
    invalid_mask.sum()
)


obe_cleaned = (
    obe_cleaned[
        ~invalid_mask
    ]
    .reset_index(drop=True)
    .copy()
)
```

```python
# ============================================================
# DETECT INTERPOLATED ORDINAL VALUES
# ============================================================

ordinal_features = [
    'FCVC',
    'NCP',
    'CH2O',
    'FAF',
    'TUE'
]


interpolated_count = np.zeros(
    len(obe_cleaned),
    dtype=int
)


for column in ordinal_features:

    is_interpolated = ~np.isclose(

        obe_cleaned[column],

        np.round(
            obe_cleaned[column]
        )

    )

    interpolated_count += (
        is_interpolated.astype(int)
    )


obe_cleaned[
    'Interpolated_Count'
] = interpolated_count
```

```python
# ============================================================
# INSPECT FILTERING RULE
# ============================================================

rows_to_remove = (
    obe_cleaned[
        'Interpolated_Count'
    ] >= 3
)


print(
    "Rows Before Filtering:",
    len(obe_cleaned)
)


print(
    "Rows Identified For Removal:",
    rows_to_remove.sum()
)


print(
    "Rows Remaining:",
    (~rows_to_remove).sum()
)
```

```python
# ============================================================
# REMOVE HIGHLY INTERPOLATED RECORDS
# ============================================================

obe_cleaned = (

    obe_cleaned[

        obe_cleaned[
            'Interpolated_Count'
        ] < 3

    ]

    .drop(
        columns=[
            'Interpolated_Count'
        ]
    )

    .reset_index(drop=True)

    .copy()
)


print(
    "Final Dataset Shape:",
    obe_cleaned.shape
)
```

```python
# ============================================================
# CHECK CLASS DISTRIBUTION
# ============================================================

class_distribution = (

    obe_cleaned[
        'NObeyesdad'
    ]

    .value_counts()

)


print(
    class_distribution
)
```

```python
# ============================================================
# CLASS DISTRIBUTION VISUALIZATION
# ============================================================

import matplotlib.pyplot as plt


class_distribution.plot(
    kind='bar',
    figsize=(10, 6)
)


plt.title(
    'Class Distribution After Data Preparation'
)

plt.xlabel(
    'Obesity Level'
)

plt.ylabel(
    'Number of Records'
)

plt.xticks(
    rotation=45,
    ha='right'
)

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# TARGET ENCODING
# ============================================================

from sklearn.preprocessing import LabelEncoder


le = LabelEncoder()


obe_cleaned[
    'Target'
] = le.fit_transform(
    obe_cleaned[
        'NObeyesdad'
    ]
)


obesity_mapping = dict(

    zip(

        le.classes_,

        le.transform(
            le.classes_
        )

    )

)


print(
    obesity_mapping
)
```

```python
# ============================================================
# SEPARATE FEATURES AND TARGET
# ============================================================

X = obe_cleaned.drop(
    columns=[
        'NObeyesdad',
        'Target'
    ]
)


y = obe_cleaned[
    'Target'
]
```

```python
# ============================================================
# ONE-HOT ENCODING
# ============================================================

X_encoded = pd.get_dummies(
    X,
    drop_first=True,
    dtype=int
)


print(
    "Encoded Dataset Shape:",
    X_encoded.shape
)


display(
    X_encoded.head()
)
```

```python
# ============================================================
# TRAIN / TEST SPLIT
# ============================================================

from sklearn.model_selection import train_test_split


X_train, X_test, y_train, y_test = train_test_split(

    X_encoded,

    y,

    test_size=0.20,

    random_state=42,

    stratify=y
)


print(
    "Training Set:",
    X_train.shape
)


print(
    "Testing Set:",
    X_test.shape
)
```

```python
# ============================================================
# STANDARDIZATION
# ============================================================

from sklearn.preprocessing import StandardScaler


scaler = StandardScaler()


X_train_scaled = scaler.fit_transform(
    X_train
)


X_test_scaled = scaler.transform(
    X_test
)
```

```python
# ============================================================
# MODEL 1 - DECISION TREE
# ============================================================

dt_model = DecisionTreeClassifier(
    random_state=42
)

dt_param_grid = {
    'criterion': [
        'gini',
        'entropy'
    ],

    'max_depth': [
        3,
        5,
        7,
        10
    ],

    'min_samples_split': [
        5,
        10,
        20
    ],

    'min_samples_leaf': [
        3,
        5,
        10
    ]
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

dt_grid = GridSearchCV(
    estimator=dt_model,
    param_grid=dt_param_grid,
    scoring='f1_macro',
    cv=cv,
    n_jobs=-1
)

dt_grid.fit(
    X_train_scaled,
    y_train
)


best_dt = (
    dt_grid.best_estimator_
)


dt_pred = best_dt.predict(
    X_test_scaled
)
```

```python
# ============================================================
# MODEL 2 - Logistic Regression (Baseline)
# ============================================================

lr_model = LogisticRegression(
    C=1.0,
    solver='lbfgs',
    max_iter=3000,
    penalty='l2',
    random_state=42
)

lr_model.fit(
    X_train_scaled,
    y_train
)


lr_pred = lr_model.predict(
    X_test_scaled
)

best_lr = lr_model
```

```python
# ============================================================
# MODEL 3 - K-NEAREST NEIGHBORS
# ============================================================

knn_model = KNeighborsClassifier()


knn_param_grid = {
    'n_neighbors': [3, 5, 7, 9, 11, 15, 21, 25],
    'weights': ['uniform', 'distance'],
    'metric': ['euclidean', 'manhattan', 'minkowski'],
    'p': [1, 2]
}


knn_grid = GridSearchCV(
    estimator=knn_model,
    param_grid=knn_param_grid,
    scoring='accuracy',
    cv=cv,
    n_jobs=-1
)

knn_grid.fit(
    X_train_scaled,
    y_train
)


best_knn = (
    knn_grid.best_estimator_
)


knn_pred = best_knn.predict(
    X_test_scaled
)
```

```python
# ============================================================
# MODEL EVALUATION
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)


def evaluate_model(
    model_name,
    y_true,
    y_pred
):

    return {

        'Model':
            model_name,

        'Accuracy':
            accuracy_score(
                y_true,
                y_pred
            ),

        'Macro Precision':
            precision_score(
                y_true,
                y_pred,
                average='macro',
                zero_division=0
            ),

        'Macro Recall':
            recall_score(
                y_true,
                y_pred,
                average='macro',
                zero_division=0
            ),

        'Macro F1-Score':
            f1_score(
                y_true,
                y_pred,
                average='macro',
                zero_division=0
            )
    }
```

```python
# ============================================================
# DISPLAY BEST PARAMETERS
# ============================================================

print("\n")
print("=" * 75)
print("BEST PARAMETERS FROM GRIDSEARCHCV")
print("=" * 75)


print("\nDecision Tree:")
print(dt_grid.best_params_)

print(
    "CV Macro F1:",
    f"{dt_grid.best_score_ * 100:.2f}%"
)


print("\nLogistic Regression (Baseline):")

print({
    "C": lr_model.C,
    "solver": lr_model.solver,
    "penalty": lr_model.penalty,
    "max_iter": lr_model.max_iter
})

print(
    "Tuning:",
    "Fixed parameters / No GridSearchCV"
)


print("\nK-Nearest Neighbors:")
print(knn_grid.best_params_)

print(
    "CV Macro F1:",
    f"{knn_grid.best_score_ * 100:.2f}%"
)
```

```python
# ============================================================
# EVALUATE BEST MODELS
# ============================================================

dt_result = evaluate_model(
    'Decision Tree',
    y_test,
    dt_pred
)


lr_result = evaluate_model(
    'Logistic Regression',
    y_test,
    lr_pred
)


knn_result = evaluate_model(
    'K-Nearest Neighbors',
    y_test,
    knn_pred
)


results_df = pd.DataFrame([
    dt_result,
    lr_result,
    knn_result
])


display(results_df)
```

```python
# ============================================================
# OVERALL PERFORMANCE COMPARISON GRAPH
# ============================================================

plot_metrics = [
    'Accuracy',
    'Macro Precision',
    'Macro Recall',
    'Macro F1-Score'
]


plot_df = results_df[
    ['Model'] + plot_metrics
].copy()


# Convert to percentage
for col in plot_metrics:

    plot_df[col] = (
        plot_df[col] * 100
    )


plot_df = plot_df.set_index(
    'Model'
)


ax = plot_df.plot(
    kind='bar',
    figsize=(13, 7),
    width=0.75
)


plt.title(
    'Overall Performance Comparison of Machine Learning Models',
    fontsize=15,
    pad=15
)


plt.xlabel(
    'Machine Learning Model',
    fontsize=12
)


plt.ylabel(
    'Performance Score (%)',
    fontsize=12
)


plt.ylim(
    0,
    105
)


plt.xticks(
    rotation=0,
    fontsize=10
)


plt.legend(
    title='Evaluation Metrics',
    loc='lower right'
)


# Add percentage labels
for container in ax.containers:

    labels = []

    for bar in container:

        height = bar.get_height()

        labels.append(
            f'{height:.2f}%'
        )

    ax.bar_label(
        container,
        labels=labels,
        padding=3,
        fontsize=9
    )


plt.tight_layout()

plt.show()
```
